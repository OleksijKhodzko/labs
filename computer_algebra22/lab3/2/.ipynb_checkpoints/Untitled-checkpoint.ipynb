```python
%matplotlib ipympl
#%matplotlib widget
#%matplotlib notebook
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.animation import FuncAnimation
```

```python
q1 = 1
m1 = 1
q2 = -1
m2 = 1
A = [2, 1]
B = [5, 4]
```

```python
def rect(x, B):
    B = int(B)
    x = int(x)

    high = np.ones(B)
    low1 = np.zeros(int(x/2 - B/2))
    x1 = np.append(low1, high)
    rect = np.append(x1, low1)

    if x > len(rect):
        rect = np.append(rect, 0)
    elif x <  len(rect):
        rect = rect[:-1]
    return rect
```

```python
pi = np.pi
length = 2000
x = np.linspace(-1, 1, length)

fig, (ax1, ax2) = plt.subplots(2, figsize=(12,6))
sinc, = ax1.plot(x, np.sin(x), '-b')
box, = ax2.plot(x, np.sin(x), '-r')

def animate(B):
    f = rect(len(x), B)
    box.set_ydata(f)

    F = (B/ length) * np.sin(x *B / 2) / (x * B / 2)
    sinc.set_ydata(F)
anim_running = True

def onClick(event):
    global anim_running
    if anim_running:
        ani.event_source.stop()
        anim_running = False
    else:
        ani.event_source.start()
        anim_running = True


def init():
    
    
    ax2.set_ylim(-0.2, 1.1)
    ax1.set_xlim(-0.25, 0.25)
    ax2.set_xlim(-1, 1)
    ax1.axhline(0, color='black', lw=1)
    ax2.axhline(0, color='black', lw=1)
    plt.rcParams.update({'font.size': 10})

    return sinc, box


step = 10
steps = np.arange(10, 1000, step)
ani = FuncAnimation(fig, animate, steps, init_func=init, interval=3, blit=True)
fig.canvas.mpl_connect('button_press_event', onClick)

plt.show()
```
